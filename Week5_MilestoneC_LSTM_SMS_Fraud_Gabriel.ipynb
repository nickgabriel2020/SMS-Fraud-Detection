```python
######################################################################
# DS-502 Predictive Analytics                                        #
# Week 5 - Milestone C                                               #
# Prototype Implementation, Diagnostics & Progress Reports           #
# SMS Fraud Detection Using Long Short-Term Memory (LSTM)            #
# Nick Gabriel                                             July 2026 #                                                         #
######################################################################

######################################################################
# This program implements and evaluates a Long Short-Term Memory     #
# (LSTM) neural network for SMS fraud detection.                     #
#                                                                    #
# The program continues the work completed during Milestone B by     #
# constructing, training, and evaluating an initial deep learning    #
# prototype.                                                         #
#                                                                    #
# This project performs:                                             #
# - Dataset loading                                                  #
# - Data preprocessing                                               #
# - Exploratory data analysis                                        #
# - Feature engineering                                              #
# - Feature encoding                                                 #
# - Dataset preparation                                              #
# - LSTM model construction                                          #
# - Model training                                                   #
# - Model evaluation                                                 #
# - Performance visualization                                        #
######################################################################


######################################################################
# 1. Import Required Libraries                                       #
# Import the Python libraries required for data loading,             #
# preprocessing, visualization, exploratory analysis, and text       #
# preparation for deep learning.                                     #
######################################################################
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing   import LabelEncoder
from sklearn.model_selection import train_test_split


######################################################################
# TensorFlow / Keras Libraries                                       #
# Import the TensorFlow and Keras libraries required to build,       #
# train, and evaluate the Long Short-Term Memory (LSTM) neural       #
# network used for SMS fraud detection.                              #
######################################################################

from tensorflow.keras.models import Sequential

from tensorflow.keras.layers import (Input, LSTM, Dense, Dropout)

from sklearn.metrics import (
    confusion_matrix,
    classification_report)


######################################################################
# SMS Fraud Detection Using Long Short-Term Memory (LSTM)            #
# Week 5 - Milestone C                                               #
# Prototype Implementation and Initial Model Evaluation              #
######################################################################

print("=" * 70)
print("SMS FRAUD DETECTION USING LONG SHORT-TERM MEMORY (LSTM)")
print("Week 5 - Milestone C")
print("Prototype Implementation and Initial Model Evaluation")
print("=" * 70, "\n")

print("1. Library modules have been imported successfully.\n")


######################################################################
# 2. Configure Display Options                                       #
# Configure pandas display settings to improve dataset readability   #
# during exploration and preprocessing.                              #
######################################################################
pd.set_option("display.max_rows",    None)
pd.set_option("display.max_columns", None)
pd.set_option("display.width",       None)
print("2. Display options have been configured.\n")

######################################################################
# Plot Color Definitions                                             #
# Define a consistent color scheme for all visualizations used       #
# throughout this notebook.                                          #
######################################################################
legitimateColor = "forestgreen" # slightly ligher green
fraudulentColor = "crimson"     # slightly darker red
graphColor      = "royalblue"

######################################################################
# 3. Load SMS Fraud Dataset                                          #
# Load the SMS Fraud Detection dataset into a pandas DataFrame so    #
# that it can be inspected, cleaned, and prepared for deep learning. #
######################################################################
# dataFilePath = r"F:\PythonDataFiles\SMS-Fraud\SMS-Fraud-Data.csv"
dataFilePath = "SMS-Fraud-Data.csv"

print("3. Loading SMS Fraud Detection Dataset SMS-Fraud-Data.csv ...")

smsDataFrame = pd.read_csv(dataFilePath)
print("   Dataset SMS-Fraud-Data.csv has been loaded successfully.\n")

######################################################################
# 4. Dataset Preview                                                 #
# Display a preview of the dataset, including sample records,        #
# dataset dimensions, and variable information to verify that the    #
# dataset loaded correctly.                                          #
######################################################################
print("4A. Dataset Preview (First Five Records)\n")

print(smsDataFrame.head(), "\n")

print("4B. Dataset Shape=", smsDataFrame.shape, "\n")

print("4C. Dataset Information\n")

smsDataFrame.info()
print()

######################################################################
# 5. Optional Dataset Sampling                                       #
# During development, a smaller sample may be used to speed up       #
# testing. The full dataset should be used for the final notebook.   #
######################################################################
# SampleData = True
SampleData = False

if SampleData:

    smsDataFrame = smsDataFrame.sample(
        n=10000,
        random_state=50)

    print("5. DEVELOPMENT MODE")
    print("   Sample dataset selected.\n")

else:
    print("5. FULL DATASET selected for analysis.\n")

######################################################################
# 6. Display Dataset Column Names                                    #
# Display all dataset column names to verify the available           #
# variables and identify the features that may contribute to SMS     #
# fraud detection.                                                   #
######################################################################
print("6. Dataset Column Names\n")

for columnNumber, columnName in enumerate(
        smsDataFrame.columns,
        start=1):

    print(columnNumber, columnName)

print()

######################################################################
# 7. Missing Value Analysis                                          #
# Examine the dataset for missing values to determine which          #
# variables may require cleaning before deep learning preprocessing. #
######################################################################
print("7. Missing Value Analysis\n")

missingValueCounts = smsDataFrame.isnull().sum()

print(missingValueCounts, "\n")

missingValuePercentages = (smsDataFrame.isnull().mean() * 100)

print("Missing Value Percentages (%)\n")

print(missingValuePercentages.sort_values(ascending=False), "\n")

######################################################################
# 8. Missing Value Visualization                                     #
# Display the percentage of missing values for each feature to       #
# identify variables requiring preprocessing before model design.    #
######################################################################
print("8. Missing Value Visualization\n")

fig = plt.figure("8. Missing Values")

ax = fig.add_subplot(111)

missingValuePercentages.sort_values(
    ascending=False).plot(kind="bar", color=graphColor, ax=ax)

ax.set_title(
    "8. Missing Values by Feature\n"
    "Percentage of Missing Values")

ax.set_ylabel("Percent Missing")

plt.tight_layout()

plt.show()

######################################################################
# 9. Duplicate Record Analysis                                       #
# Identify duplicate observations that may negatively influence      #
# model training and evaluation. Remove duplicate records and        #
# summarize the results.                                             #
######################################################################
print("9. Duplicate Record Analysis\n")

# Count duplicate records
duplicateCount = smsDataFrame.duplicated().sum()
print("Duplicate Records Found :", duplicateCount, "\n")

# Store original record count
originalRecordCount = len(smsDataFrame)

# Remove duplicates
smsDataFrame = smsDataFrame.drop_duplicates()

# Count remaining records
remainingRecordCount = len(smsDataFrame)

removedRecordCount = originalRecordCount - remainingRecordCount

print("=" * 60)
print("Observation Summary")
print("=" * 60)

print("Original Number of Records :", originalRecordCount)

print("Duplicate Records Removed  :", removedRecordCount)

print("Remaining Number of Records:", remainingRecordCount, "\n")

######################################################################
# 10. Missing Value Replacement                                      #
# Replace missing numeric values using the median and categorical    #
# values using the mode. Summarize the number of missing values      #
# before and after preprocessing.                                    #
######################################################################
print("10. Missing Value Replacement\n")

# Count missing values before cleaning
missingValuesBefore = (
    smsDataFrame
    .isnull()
    .sum()
    .sum())

print("Missing Values Before Cleaning :", missingValuesBefore, "\n")

# Replace numeric missing values

numericColumns = (
    smsDataFrame
    .select_dtypes(include=np.number)
    .columns)

for columnName in numericColumns:
    smsDataFrame[columnName] = (
        smsDataFrame[columnName]
        .fillna(
            smsDataFrame[columnName].median()
        )
    )

# Replace categorical missing values
categoricalColumns = (
    smsDataFrame
    .select_dtypes(exclude=np.number)
    .columns)

for columnName in categoricalColumns:
    smsDataFrame[columnName] = (
        smsDataFrame[columnName]
        .fillna(
            smsDataFrame[columnName].mode()[0]
        )
    )

# Count missing values after cleaning

missingValuesAfter = (
    smsDataFrame
    .isnull()
    .sum()
    .sum())

missingValuesReplaced = (
    missingValuesBefore -
    missingValuesAfter)

print("=" * 60)
print("Observation Summary")
print("=" * 60)

print("Missing Values Before Cleaning :", missingValuesBefore)

print("Missing Values Replaced        :", missingValuesReplaced)

print("Missing Values Remaining       :", missingValuesAfter, "\n")

######################################################################
# 11. Dataset Dimensions                                             #
# Display the total number of records and features to understand     #
# the overall size of the dataset before deep learning preparation.  #
######################################################################
print("11. Dataset Dimensions\n")

numberOfRows = smsDataFrame.shape[0]
numberOfColumns = smsDataFrame.shape[1]

print("Number of Records :", numberOfRows)
print("Number of Features:", numberOfColumns, "\n")


######################################################################
# 12. Data Type Summary                                              #
# Summarize the dataset data types to understand which variables     #
# are numeric and which are categorical.                             #
######################################################################
print("12. Data Type Summary\n")

print(smsDataFrame.dtypes, "\n")

print("Data Type Counts")
print(smsDataFrame.dtypes.value_counts(), "\n")

######################################################################
# 13. Label Distribution                                             #
# Examine the distribution of legitimate and fraudulent SMS          #
# messages to determine whether class imbalance exists.              #
######################################################################
print("13. Label Distribution\n")

print(smsDataFrame["label"].value_counts(), "\n")

fig = plt.figure("13. Label Distribution")

ax = fig.add_subplot(111)

sns.countplot(
    data=smsDataFrame,
    x="label",
    hue="label",
    palette={
        "legitimate": legitimateColor,
        "fraudulent": fraudulentColor
    },
    legend=False,
    ax=ax)

ax.set_title(
    "13. Label Distribution\n"
    "Distribution of Legitimate and Fraudulent Messages")

ax.set_xlabel("Message Label")
ax.set_ylabel("Count")

plt.tight_layout()

plt.show()


######################################################################
# 14. Sender Type Distribution                                       #
# Examine the different sender types represented in the dataset      #
# and determine how frequently each sender type occurs.              #
######################################################################
print("14. Sender Type Distribution\n")

print(smsDataFrame["sender_type"].value_counts(dropna=False), "\n")

fig = plt.figure("14. Sender Type Distribution")

ax = fig.add_subplot(111)

smsDataFrame["sender_type"].value_counts(
    dropna=False).plot(
    kind="bar",
    color=graphColor,
    ax=ax)

ax.set_title("14. Sender Type Distribution")

ax.set_ylabel("Number of Messages")
plt.tight_layout()
plt.show()

######################################################################
# 15. Organization Distribution                                      #
# Identify the organizations that appear most frequently in the      #
# dataset. Displaying the most common organizations helps reveal     #
# which businesses or institutions are most commonly represented     #
# within the SMS Fraud dataset.                                      #
######################################################################
print("15. Organization Distribution\n")

organizationCounts = (
    smsDataFrame["organization_name"]
    .value_counts())

print(organizationCounts, "\n")

# Display only the Top 15 organizations
topOrganizations = organizationCounts.head(15)

fig = plt.figure(
    "15. Organization Distribution",
    figsize=(12, 8))

ax = fig.add_subplot(111)

topOrganizations.sort_values().plot(
    kind="barh",
    color=graphColor,
    ax=ax)

ax.set_title("15. Which Organizations Appear Most Frequently?")
ax.set_xlabel("Number of Messages")
ax.set_ylabel("Organization")
plt.tight_layout()

plt.show()

print("=" * 60)
print("Observation Summary")
print("=" * 60)

print("The organization with the largest number of messages is:",
    organizationCounts.idxmax())

print("Total Messages:", organizationCounts.max(), "\n")

######################################################################
# 16. Message Template Distribution                                  #
# Examine the frequency of the message templates contained in the    #
# dataset. Understanding common templates provides insight into      #
# recurring SMS communication patterns.                              #
######################################################################
print("16. Message Template Distribution\n")

messageTemplateCounts = (
    smsDataFrame["message_template"]
    .value_counts())

print(messageTemplateCounts, "\n")

# Display only the Top 15 templates in the graph
topMessageTemplates = messageTemplateCounts.head(15)

fig = plt.figure(
    "16. Message Template Distribution",
    figsize=(12, 8))

ax = fig.add_subplot(111)

topMessageTemplates.plot(kind="bar", color=graphColor, ax=ax)

ax.set_title("16. Which Message Templates Appear Most Frequently?")

ax.set_xlabel("Message Template")

ax.set_ylabel("Number of Messages")

plt.xticks(rotation=90, fontsize=9)

plt.tight_layout()

plt.show()

######################################################################
# 17. Communication Channel Distribution                             #
# Examine the communication channels represented in the dataset.     #
# Although the current dataset contains only SMS messages, this      #
# feature supports future datasets that may include additional       #
# messaging platforms such as MMS, WhatsApp, RCS, or email.          #
######################################################################
print("17. Communication Channel Distribution\n")

channelCounts = (
    smsDataFrame["delivery_method"]
    .value_counts())

print(channelCounts, "\n")

mostCommonChannel = channelCounts.idxmax()

mostCommonCount = channelCounts.max()

mostCommonPercent = mostCommonCount / channelCounts.sum() * 100

fig = plt.figure(
    "17. Communication Channel Distribution",
    figsize=(6,6))

ax = fig.add_subplot(111)

channelCounts.plot(
    kind="pie",
    autopct="%1.1f%%",
    startangle=90,
    color=graphColor,
    ax=ax)

ax.set_title(
    "17. Communication Channels Represented\n"
    "in the SMS Fraud Dataset")

ax.set_ylabel("")

plt.tight_layout()

plt.show()

print("=" * 60)
print("Observation Summary")
print("=" * 60)

print("Primary Communication Channel :", mostCommonChannel)

print("Number of Messages            :", mostCommonCount)

print("Percentage of Dataset         :", format(mostCommonPercent, ".2f"), "\n")

print("Observation: The current dataset contains SMS messages only.\n"
    "The communication channel feature has been retained because\n"
    "future versions of the dataset may include additional messaging\n"
    "platforms such as MMS, WhatsApp, RCS, or email.", "\n")

######################################################################
# 18. Transaction Amount Statistics                                  #
# Summarize the transaction amounts contained in the dataset.        #
######################################################################
print("18. Transaction Amount Statistics\n")

print(smsDataFrame["transaction_amount"].describe(), "\n")

######################################################################
# 19. Transaction Amount Distribution                                #
# Visualize the distribution of transaction amounts represented      #
# in the dataset. This histogram illustrates how transaction values  #
# are distributed across all SMS messages.                           #
######################################################################
print("19. Transaction Amount Distribution\n")

fig = plt.figure(
    "19. Transaction Amount Distribution")

ax = fig.add_subplot(111)

smsDataFrame["transaction_amount"].plot(
    kind="hist",
    bins=30,
    color=graphColor,
    ax=ax)

ax.set_title("19. Distribution of Transaction Amounts")
ax.set_xlabel("Transaction Amount")
ax.set_ylabel("Number of Messages")

plt.tight_layout()

plt.show()

print("=" * 60)
print("Observation Summary")
print("=" * 60)

print("Minimum Transaction Amount :",
      smsDataFrame["transaction_amount"].min())

print("Maximum Transaction Amount :",
      smsDataFrame["transaction_amount"].max())

print("Average Transaction Amount :",
     format(smsDataFrame['transaction_amount'].mean(),
            ".2f"), "\n")

######################################################################
# 20. Currency Distribution                                          #
# Examine the currencies referenced throughout the dataset. This     #
# visualization illustrates the proportion of messages associated    #
# with each currency.                                                #
######################################################################
print("20. Currency Distribution\n")

currencyCounts = (
    smsDataFrame["currency"]
    .value_counts())

print(currencyCounts, "\n")

mostCommonCurrency = currencyCounts.idxmax()

mostCommonCount = currencyCounts.max()

mostCommonPercent = mostCommonCount / currencyCounts.sum() * 100

fig = plt.figure("20. Currency Distribution", figsize=(6,6))

ax = fig.add_subplot(111)

currencyCounts.plot(
    kind="pie",
    autopct="%1.1f%%",
    startangle=90,
    color=graphColor,
    ax=ax)

ax.set_title(
    "20. Distribution of Transaction Currencies\n"
    "Most Transactions are Denominated in "
    + mostCommonCurrency)

ax.set_ylabel("")

plt.tight_layout()

plt.show()

print("=" * 60)
print("Observation Summary")
print("=" * 60)

print("Most Frequently Referenced Currency :", mostCommonCurrency)

print("Number of Messages                  :", mostCommonCount)

print("Percentage of Dataset               :",
      format(mostCommonPercent, ".2f"), "\n")

######################################################################
# 21. Message Length Category Distribution                           #
# Examine the distribution of message length categories contained    #
# in the dataset. These categories summarize the approximate length  #
# of each SMS message and provide insight into message size          #
# characteristics without requiring the original message text.       #
######################################################################
print("21. Message Length Category Distribution\n")

lengthCategoryCounts = (
    smsDataFrame["message_length_category"]
    .value_counts(dropna=False))

print(lengthCategoryCounts, "\n")


fig = plt.figure(
    "21. Message Length Category Distribution",
    figsize=(8, 6))

ax = fig.add_subplot(111)

lengthCategoryCounts.plot(kind="bar", color=graphColor, ax=ax)

ax.set_title("21. Distribution of Message Length Categories")

ax.set_xlabel("Message Length Category")

ax.set_ylabel("Number of Messages")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()


mostCommonCategory = (lengthCategoryCounts.idxmax())

mostCommonCount = (lengthCategoryCounts.max())

percentageOfDataset = (
    mostCommonCount /
    lengthCategoryCounts.sum()) * 100

print("=" * 60)
print("Observation Summary")
print("=" * 60)

print("Most Common Message Length Category :", mostCommonCategory)

print("Number of Messages                  :", mostCommonCount)

print("Percentage of Dataset               :",
    format(percentageOfDataset, ".2f"), "\n")

######################################################################
# 22. Message Tone Distribution                                      #
# Examine the message tones represented in the dataset. Message      #
# tone may provide useful information for distinguishing between     #
# legitimate and fraudulent SMS messages.                            #
######################################################################
print("22. Message Tone Distribution\n")

messageToneCounts = (
    smsDataFrame["message_tone"]
    .value_counts(dropna=False))

print(messageToneCounts, "\n")

fig = plt.figure(
    "22. Message Tone Distribution",
    figsize=(8,6))

ax = fig.add_subplot(111)

messageToneCounts.plot(kind="bar", color=graphColor, ax=ax)

ax.set_title("22. Distribution of Message Tones")

ax.set_xlabel("Message Tone")

ax.set_ylabel("Number of Messages")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()

mostCommonTone = (messageToneCounts.idxmax())

mostCommonCount = (messageToneCounts.max())

percentageOfDataset = (
    mostCommonCount /
    messageToneCounts.sum()) * 100

print("=" * 60)
print("Observation Summary")
print("=" * 60)

print("Most Common Message Tone :", mostCommonTone)

print("Number of Messages       :", mostCommonCount)

print("Percentage of Dataset    :",
      format(percentageOfDataset, ".2f"), "\n")

######################################################################
# 23. Message Tone by Label                                          #
# Compare message tones across legitimate and fraudulent messages    #
# to determine whether certain tones are more frequently associated  #
# with fraudulent SMS messages.                                      #
######################################################################
print("23. Message Tone by Label\n")

messageToneByLabel = pd.crosstab(
    smsDataFrame["message_tone"],
    smsDataFrame["label"])

print(messageToneByLabel, "\n")

fig = plt.figure("23. Message Tone by Label", figsize=(10,6))

ax = fig.add_subplot(111)

messageToneByLabel = messageToneByLabel[["legitimate", "fraudulent"]]

messageToneByLabel.plot(
    kind="bar",
    ax=ax,
    color=[legitimateColor, fraudulentColor])

ax.set_title("23. Message Tone by Message Label")

ax.set_xlabel("Message Tone")

ax.set_ylabel("Number of Messages")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()


print("=" * 60)
print("Observation Summary")
print("=" * 60)

print("The distribution above illustrates how message tones")

print("are associated with legitimate and fraudulent SMS")

print("messages within the dataset.", "\n")

######################################################################
# 24. Call-to-Action Distribution                                    #
# Examine the different call-to-action types represented in the      #
# dataset. Fraudulent messages frequently encourage recipients to    #
# perform immediate actions, making this feature potentially useful  #
# for classification.                                                #
######################################################################

print("24. Call-to-Action Distribution\n")

callToActionCounts = (
    smsDataFrame["call_to_action"]
    .value_counts(dropna=False))

print(callToActionCounts, "\n")

fig = plt.figure(
    "24. Call-to-Action Distribution",
    figsize=(8,6))

ax = fig.add_subplot(111)

callToActionCounts.plot(
    kind="bar",
    color=graphColor,
    ax=ax)

ax.set_title("24. Distribution of Call-to-Action Types")

ax.set_xlabel("Call-to-Action")

ax.set_ylabel("Number of Messages")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()

mostCommonAction = (callToActionCounts.idxmax())

mostCommonCount = (callToActionCounts.max())

percentageOfDataset = (mostCommonCount /
    callToActionCounts.sum()) * 100

print("=" * 60)
print("Observation Summary")
print("=" * 60)

print("Most Common Call-to-Action :", mostCommonAction)

print("Number of Messages         :", mostCommonCount)

print("Percentage of Dataset      :",
    format(percentageOfDataset, ".2f"), "\n")

######################################################################
# 25. Call-to-Action by Label                                        #
# Compare call-to-action types across legitimate and fraudulent      #
# messages to determine whether certain actions are more commonly    #
# associated with fraud.                                             #
######################################################################

print("25. Call-to-Action by Label\n")

callToActionByLabel = pd.crosstab(
    smsDataFrame["call_to_action"],
    smsDataFrame["label"])

print(callToActionByLabel, "\n")


fig = plt.figure("25. Call-to-Action by Label", figsize=(10,6))

ax = fig.add_subplot(111)

callToActionByLabel = callToActionByLabel[["legitimate", "fraudulent"]]

callToActionByLabel.plot(
    kind="bar",
    ax=ax,
    color=[legitimateColor, fraudulentColor])

ax.set_title("25. Call-to-Action by Message Label")

ax.set_xlabel("Call-to-Action")

ax.set_ylabel("Number of Messages")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()


print("=" * 60)
print("Observation Summary")
print("=" * 60)

print("The chart compares call-to-action types")
print("between legitimate and fraudulent messages.", "\n")

######################################################################
# 26. URL Presence Distribution                                      #
# Examine whether SMS messages contain URLs. URLs are commonly used  #
# in phishing and fraud campaigns and may provide valuable           #
# information for fraud detection.                                   #
######################################################################
print("26. URL Presence Distribution\n")

urlCounts = (
    smsDataFrame["contains_url"]
    .value_counts(dropna=False))

print(urlCounts, "\n")

fig = plt.figure("26. URL Presence Distribution", figsize=(8,6))

ax = fig.add_subplot(111)

urlCounts.plot(kind="bar", color=graphColor, ax=ax)

ax.set_title("26. Presence of URLs in SMS Messages")

ax.set_xlabel("Contains URL")

ax.set_ylabel("Number of Messages")

plt.tight_layout()

plt.show()


mostCommonValue = (urlCounts.idxmax())

mostCommonCount = (urlCounts.max())

percentageOfDataset = (mostCommonCount / urlCounts.sum()) * 100

print("=" * 60)
print("Observation Summary")
print("=" * 60)

print("Most Common Value      :", mostCommonValue)

print("Number of Messages     :", mostCommonCount)

print("Percentage of Dataset  :",
      format(percentageOfDataset, ".2f"), "\n")


######################################################################
# 27. URL Presence by Label                                          #
# Compare URL usage between legitimate and fraudulent SMS messages   #
# to determine whether URLs are more frequently associated with      #
# fraudulent activity.                                               #
######################################################################
print("27. URL Presence by Label\n")

urlByLabel = pd.crosstab(
    smsDataFrame["contains_url"],
    smsDataFrame["label"])

print(urlByLabel, "\n")


fig = plt.figure(
    "27. URL Presence by Label",
    figsize=(8,6))

ax = fig.add_subplot(111)

urlByLabel = urlByLabel[["legitimate", "fraudulent"]]

urlByLabel.plot(
    kind="bar",
    ax=ax,
    color=[legitimateColor, fraudulentColor])

ax.set_title("27. URL Presence by Message Label")

ax.set_xlabel("Contains URL")

ax.set_ylabel("Number of Messages")

plt.tight_layout()

plt.show()

print("=" * 60)
print("Observation Summary")
print("=" * 60)

print("The chart compares URL usage between")
print("legitimate and fraudulent SMS messages\n")

######################################################################
# 28. Security Warning Distribution                                  #
# Examine whether messages contain security warning language.        #
# Security-related wording may provide useful information for        #
# distinguishing legitimate and fraudulent SMS messages.             #
######################################################################

print("28. Security Warning Distribution\n")

securityCounts = (
    smsDataFrame["contains_security_warning"]
    .value_counts(dropna=False))

print(securityCounts, "\n")


fig = plt.figure(
    "28. Security Warning Distribution",
    figsize=(8,6))

ax = fig.add_subplot(111)

securityCounts.plot(kind="bar", color=graphColor, ax=ax)

ax.set_title("28. Security Warning Language")

ax.set_xlabel("Contains Security Warning")

ax.set_ylabel("Number of Messages")

plt.tight_layout()

plt.show()


mostCommonValue = (securityCounts.idxmax())

mostCommonCount = (securityCounts.max())

percentageOfDataset = mostCommonCount / securityCounts.sum() * 100

print("=" * 60)
print("Observation Summary")
print("=" * 60)

print("Most Common Value      :", mostCommonValue)

print("Number of Messages     :", mostCommonCount)

print("Percentage of Dataset  :",
      format(percentageOfDataset, ".2f"), "\n")


######################################################################
# 29. Security Warning by Label                                      #
# Compare the presence of security warning language between          #
# legitimate and fraudulent messages.                                #
######################################################################

print("29. Security Warning by Label\n")

securityByLabel = pd.crosstab(
    smsDataFrame["contains_security_warning"],
    smsDataFrame["label"])

print(securityByLabel, "\n")


fig = plt.figure(
    "29. Security Warning by Label",
    figsize=(8,6))

ax = fig.add_subplot(111)

securityByLabel = securityByLabel[["legitimate", "fraudulent"]]

securityByLabel.plot(
    kind="bar",
    ax=ax,
    color=[legitimateColor, fraudulentColor])

ax.set_title("29. Security Warning by Message Label")

ax.set_xlabel("Contains Security Warning")

ax.set_ylabel("Number of Messages")

plt.tight_layout()

plt.show()


print("=" * 60)
print("Observation Summary")
print("=" * 60)

print("The chart compares security warning language")
print("across legitimate and fraudulent SMS messages.", "\n")



######################################################################
# 30. Label Encoding                                                 #
# Convert the target class labels into numeric values suitable for   #
# deep learning model training.                                      #
######################################################################

print("30. Label Encoding\n")

labelEncoder = LabelEncoder()

smsDataFrame["encoded_label"] = (
    labelEncoder.fit_transform(
        smsDataFrame["label"]))

print("Label Encoding Mapping\n")

encodingMapping = pd.DataFrame(
    {
        "Original Label": labelEncoder.classes_,
        "Encoded Value": range(len(labelEncoder.classes_))
    })

print(encodingMapping, "\n")

print("Encoded Label Distribution\n")

print(smsDataFrame["encoded_label"].value_counts(), "\n")

######################################################################
# 31. Encode Categorical Features                                    #
# Convert the time of day into an hourly feature, remove identifier  #
# fields, and convert categorical predictor variables into numeric   #
# format using one-hot encoding.                                     #
######################################################################

print("31. Encode Categorical Features\n")

######################################################################
# Convert Time of Day to Hour                                        #
# Convert the original timestamp into an hourly value to reduce      #
# the number of predictor variables and improve model efficiency.    #
######################################################################

smsDataFrame["messageHour"] = pd.to_datetime(
    smsDataFrame["time_of_day"],
    format="%I:%M %p").dt.hour

######################################################################
# Remove Identifier Columns                                          #
######################################################################

featureDataFrame = smsDataFrame.drop(
    columns=[
        "label",
        "encoded_label",
        "originating_phone_number",
        "target_phone_number",
        "callback_number",
        "invoice_number",
        "transaction_id",
        "time_of_day"
    ]
)

######################################################################
# One-Hot Encode Categorical Features                                #
######################################################################

featureDataFrame = pd.get_dummies(
    featureDataFrame,
    drop_first=True)

print("Encoded Feature Matrix Shape=", featureDataFrame.shape, "\n")

print("First Five Encoded Records\n")

print(featureDataFrame.head(), "\n")


######################################################################
# 32. Feature Matrix and Target Variable                             #
# Separate the encoded predictor variables from the encoded target   #
# variable in preparation for deep learning model development.       #
######################################################################

print("32. Feature Matrix and Target Variable\n")

X = featureDataFrame

y = smsDataFrame["encoded_label"]

print("Predictor Matrix Shape      = ", X.shape)
print("Target Vector Shape         = ", y.shape)
print("Number of Predictor Features= ", X.shape[1])
print("Number of Records           = ", X.shape[0], "\n")

######################################################################
# 33. Convert Predictor Matrix to Float32                           #
# Convert all predictor variables to a single numeric data type      #
# required for efficient TensorFlow and LSTM processing.             #
######################################################################

print("33. Convert Predictor Matrix to Float32\n")

X = X.astype(np.float32)

print("Predictor matrix converted to float32.\n")


######################################################################
# 34. Split the Dataset                                              #
# Divide the dataset into training and testing sets for model        #
# development and evaluation.                                        #
######################################################################

print("34. Split the Dataset\n")

XTrain, XTest, yTrain, yTest = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y)

print("Training Predictor Matrix Shape=", XTrain.shape)
print("Testing Predictor Matrix Shape =",  XTest.shape)
print("Training Target Vector Shape   =",  yTrain.shape)
print("Testing Target Vector Shape    =",  yTest.shape, "\n")

######################################################################
# 35. Reshape Data for LSTM Input                                    #
# Reshape the predictor matrices into the three-dimensional format   #
# required by an LSTM network.                                       #
######################################################################

print("35. Reshape Data for LSTM Input\n")

XTrainLSTM = np.reshape(
    XTrain.values,
    (
        XTrain.shape[0],
        1,
        XTrain.shape[1]
    )
)

XTestLSTM = np.reshape(
    XTest.values,
    (
        XTest.shape[0],
        1,
        XTest.shape[1]
    )
)

print("Training Data Shape for LSTM=", XTrainLSTM.shape)

print("Testing  Data Shape for LSTM =", XTestLSTM.shape, "\n")


######################################################################
# 36. Final Dataset Summary                                          #
# Display the final dataset characteristics after preprocessing and  #
# preparation for deep learning.                                     #
######################################################################

print("36. Final Dataset Summary\n")

print("Total Number of Records     =", len(smsDataFrame))
print("Number of Predictor Features=", X.shape[1])
print("Number of Training Records  =", len(XTrain))
print("Number of Testing Records   =", len(XTest))
print("Number of Target Classes    =", labelEncoder.classes_, "\n")

######################################################################
# 37. Planned LSTM Model Architecture                                #
# Summarize the planned deep learning model that will be developed   #
# during Milestone C.                                                #
######################################################################

print("37. Planned LSTM Model Architecture\n")

print("The prepared dataset will be used to construct an")

print("LSTM neural network during Milestone C.\n")

print("Planned model will include:\n")

print("- Input layer receiving", X.shape[1], "predictor features")

print("  - One or more LSTM hidden layers")

print("  - Dropout layer")

print("  - Dense output layer")

print("  - Sigmoid activation for binary classification\n")

print("The model will be compiled using the")

print("Adam optimizer and Binary Crossentropy loss.\n")

print("Model performance will be evaluated using")

print("Accuracy, Precision, Recall, F1-Score,")

print("and a Confusion Matrix.\n")

print("The trained model will be used to classify")
print("SMS messages as Legitimate or Fraudulent.")

######################################################################
# 38. Build the LSTM Neural Network                                  #
# Construct the Long Short-Term Memory neural network architecture   #
# used for SMS fraud classification.                                 #
######################################################################

print()
print("38. Build the LSTM Neural Network")
print()

# Create the Sequential LSTM model.
lstmModel = Sequential()

# Define the input layer.
lstmModel.add(
    Input(shape=(XTrainLSTM.shape[1], XTrainLSTM.shape[2]))
)

# Add the LSTM hidden layer.
lstmModel.add(
    LSTM(
        units=32
    )
)

# Add dropout to help reduce overfitting.
lstmModel.add(
    Dropout(rate=0.20)
)

# Add a fully connected hidden layer.
lstmModel.add(
    Dense(
        units=16,
        activation="relu"
    )
)

# Add the output layer for binary classification.
lstmModel.add(
    Dense(
        units=1,
        activation="sigmoid"
    )
)

print("LSTM neural network created successfully.", "\n")

######################################################################
# 39. Compile the LSTM Model                                         #
# Configure the LSTM neural network by specifying the optimizer,     #
# loss function, and evaluation metric used during model training.   #
######################################################################

print("39. Compile the LSTM Model\n")

# Compile the LSTM model.
lstmModel.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

print("LSTM model compiled successfully.\n")

######################################################################
# 40. Display the Model Summary                                      #
# Display a summary of the LSTM neural network architecture,         #
# including each layer, output shape, and trainable parameters.      #
######################################################################

print("40. Display the Model Summary\n")

# Display a summary of the LSTM model architecture.
lstmModel.summary()

print("\nLSTM model summary displayed successfully.\n")


######################################################################
# 41. Train the LSTM Model                                           #
# Train the LSTM model using the training dataset.                   #
######################################################################

print("41. Train the LSTM Model\n")

trainingHistory = lstmModel.fit(
    XTrainLSTM,
    yTrain,
    epochs=5,
    batch_size=32,
    validation_data=(XTestLSTM, yTest),
    verbose=1
)

print("\nLSTM model training completed successfully.\n")

######################################################################
# 42. Display Training History                                       #
# Display the training and validation metrics stored during model    #
# training for later visualization and evaluation.                   #
######################################################################

print("42. Display Training History\n")

print("Training History Keys:")
print(trainingHistory.history.keys(), "\n")

print("Training Accuracy:")
print(trainingHistory.history["accuracy"], "\n")

print("Validation Accuracy:")
print(trainingHistory.history["val_accuracy"], "\n")

print("Training Loss:")
print(trainingHistory.history["loss"], "\n")

print("Validation Loss:")
print(trainingHistory.history["val_loss"], "\n")

print("Training history displayed successfully.\n")

######################################################################
# 43. Plot Training and Validation Accuracy                          #
# Display the training and validation accuracy across all epochs.    #
######################################################################

print("43. Plot Training and Validation Accuracy")

plt.figure(figsize=(8, 5))

# Set the graph window title
plt.gcf().canvas.manager.set_window_title(
    "43. Training and Validation Accuracy")

plt.plot(
    trainingHistory.history["accuracy"],
    label="Training Accuracy",
    linewidth=2)

plt.plot(
    trainingHistory.history["val_accuracy"],
    label="Validation Accuracy",
    linewidth=2)

plt.title(
    "43. Training and Validation Accuracy",
    fontsize=14,
    fontweight="bold")

plt.xlabel("Epoch")

plt.ylabel("Accuracy")

plt.xticks(range(5), range(1, 6))

plt.grid(True)

plt.legend()

# Add explanatory note
plt.text(
    0.98,
    0.02,
    "Note: Training and validation\naccuracy overlap after\nEpoch 1, indicating\nrapid model convergence.",
    transform=plt.gca().transAxes,
    fontsize=9,
    ha="right",
    va="bottom",
    bbox=dict(facecolor="white", alpha=0.85, edgecolor="gray")
)

plt.tight_layout()

plt.show()

print()

print("Training and validation accuracy graph displayed successfully.\n")

######################################################################
# 44. Plot Training and Validation Loss                              #
# Display the training and validation loss across all epochs.        #
######################################################################

print("44. Plot Training and Validation Loss")

plt.figure(figsize=(8, 5))

# Set the graph window title
plt.gcf().canvas.manager.set_window_title(
    "44. Training and Validation Loss")

plt.plot(
    trainingHistory.history["loss"],
    label="Training Loss",
    linewidth=2)

plt.plot(
    trainingHistory.history["val_loss"],
    label="Validation Loss", linewidth=2)

plt.title(
    "44. Training and Validation Loss",
    fontsize=14,
    fontweight="bold")

plt.xlabel("Epoch")

plt.ylabel("Loss")

plt.xticks(range(5), range(1, 6))

plt.grid(True)

plt.legend()

# Add explanatory note
plt.text(
    0.98,
    0.98,
    "Note: Training and validation\nloss rapidly approach zero,\nindicating excellent model\nconvergence.",
    transform=plt.gca().transAxes,
    fontsize=9,
    ha="right",
    va="top",
    bbox=dict(facecolor="white", alpha=0.85, edgecolor="gray"))

plt.tight_layout()

plt.show()

print()

print("Training and validation loss graph displayed successfully.")
```
